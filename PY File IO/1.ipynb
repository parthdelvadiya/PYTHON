```python
# | Mode  | Meaning                  |
# | ----- | ------------------------ |
# | `"r"` | Read (default)           |
# | `"w"` | Write (overwrite)        |
# | `"a"` | Append                   |
# | `"x"` | Create (error if exists) |
# | `"b"` | Binary mode              |
# | `"t"` | Text mode                |

# r+ → read + write (no truncation)
# w+ → write + read (file overwritten)
# a+ → append + read (write at end) 
```

```python
# abcdef

f = open("file.txt", "r+")
f.write("XYZ")
f.close()

# XYZdef
```

```python
f = open("file.txt", "r+")
f.seek(0, 2)   # move pointer to end
f.write("XYZ")
f.close()

# abcdefXYZ
```

```python
with open("geek.txt", "r") as file:
    content = file.read()
    print(content)
```

```python
f = open("geek.txt", "r")
print("Filename:", f.name)
print("Mode:", f.mode)
print("Is Closed?", f.closed)

f.close()
print("Is Closed?", f.closed)

# Filename: geek.txt
# Mode: r
# Is Closed? False
# Is Closed? True
```
